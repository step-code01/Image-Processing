First read a grayscale image, and also make a simple binary (black/white only) version of it ourselves using a threshold, so we have both types of images to test on.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img_gray = cv2.imread("/content/images.jpg", cv2.IMREAD_GRAYSCALE)

if img_gray is None:
    raise FileNotFoundError("Img not found.")

plt.imshow(img_gray, cmap='gray')
plt.axis('off')
plt.title("Original grayscale image")
plt.show()
```

Now make a binary image out of the grayscale one. A binary image only has two values, 0 (black) and 255 (white).

Simple way: pick a threshold (say 127). If the pixel is greater than 127, make it 255 (white), otherwise make it 0 (black).

```python
height, width = img_gray.shape
img_binary = np.zeros((height, width), dtype=np.uint8)
threshold = 127

for i in range(height):
    for j in range(width):
        if img_gray[i][j] > threshold:
            img_binary[i][j] = 255
        else:
            img_binary[i][j] = 0

plt.imshow(img_binary, cmap='gray')
plt.axis('off')
plt.title("Binary image (made from the grayscale one)")
plt.show()
```

## Q1. Draw a border around the image (padding)

Idea:
1. Ask the user how many pixels wide the border should be, and what color the border should be.
2. Make a brand new, bigger, empty image. Its height and width are each increased by `2 * border_width` (border on all 4 sides).
3. Fill this whole new image with the border color first.
4. Then copy the original image's pixels into the middle of this new image (leaving the border untouched around it).

Do this for the grayscale image first.

```python
border_width = int(input("Enter border width in pixels: "))
border_color = int(input("Enter border color (a number from 0 to 255, 0 = black, 255 = white): "))

height, width = img_gray.shape

new_height = height + (2 * border_width)
new_width = width + (2 * border_width)

img_gray_bordered = np.zeros((new_height, new_width), dtype=np.uint8)

for i in range(new_height):
    for j in range(new_width):
        img_gray_bordered[i][j] = border_color
        
for i in range(height):
    for j in range(width):
        img_gray_bordered[i + border_width][j + border_width] = img_gray[i][j]

plt.imshow(img_gray_bordered, cmap='gray')
plt.axis('off')
plt.title("Grayscale image with border")
plt.show()
```

Now the same thing, but for the binary image. Since a binary image only understands 0 and 255, check that the user's chosen color is actually one of those two values.

```python
border_width_bin = int(input("Enter border width in pixels: "))
border_color_bin = int(input("Enter border color (0 for black or 255 for white): "))

if border_color_bin != 0 and border_color_bin != 255:
    print("Warning: a binary image only has 0 or 255 as valid colors. Using 0 instead.")
    border_color_bin = 0

height, width = img_binary.shape

new_height = height + (2 * border_width_bin)
new_width = width + (2 * border_width_bin)

img_binary_bordered = np.zeros((new_height, new_width), dtype=np.uint8)

for i in range(new_height):
    for j in range(new_width):
        img_binary_bordered[i][j] = border_color_bin

for i in range(height):
    for j in range(width):
        img_binary_bordered[i + border_width_bin][j + border_width_bin] = img_binary[i][j]

plt.imshow(img_binary_bordered, cmap='gray')
plt.axis('off')
plt.title("Binary image with border")
plt.show()
```

## Q2. Image Complement

Complement just means "invert" the image, so dark becomes light and light becomes dark.

For a grayscale image (values 0 to 255), complement of a pixel = `255 - pixel_value`.

For example: a pixel that was 0 (black) becomes 255 (white), and a pixel that was 200 becomes 55.

```python
height, width = img_gray.shape

img_gray_complement = np.zeros((height, width), dtype=np.uint8)

for i in range(height):
    for j in range(width):
        img_gray_complement[i][j] = 255 - img_gray[i][j]

plt.imshow(img_gray_complement, cmap='gray')
plt.axis('off')
plt.title("Grayscale complement")
plt.show()
```

Now do complement for the binary image. Since a binary image only has 0 and 255, complement just means swapping them: 0 becomes 255, and 255 becomes 0.

```python
height, width = img_binary.shape

img_binary_complement = np.zeros((height, width), dtype=np.uint8)

for i in range(height):
    for j in range(width):
        if img_binary[i][j] == 0:
            img_binary_complement[i][j] = 255
        else:
            img_binary_complement[i][j] = 0

plt.imshow(img_binary_complement, cmap='gray')
plt.axis('off')
plt.title("Binary complement")
plt.show()
```
